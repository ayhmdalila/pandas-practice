```python
import pandas as pd
df = pd.read_csv("IRIS.csv")
```

```python
df
```

# Groupby function & different aggregations

## Min, max

```python
df.groupby("species").aggregate("max")
```

```python
df.groupby("species").aggregate("min")
```

## Avg/mean

```python
df.groupby("species").aggregate("mean")
```

## Count

```python
df.groupby("species").aggregate("count")
```

## Sum

```python
df.groupby("species").aggregate("sum")
```

# Specifying aggregations per column

```python
df.groupby("species").aggregate({"sepal_length": "mean"})
```

```python
df.groupby("species").aggregate({"sepal_length": "mean", "sepal_length": "max"})
```

```python
df.groupby("species").aggregate({"sepal_length": "mean", "petal_length": "max"})
```
