```python
import pandas as pd
df = pd.read_csv("IRIS.csv")
```

```python
import matplotlib.pyplot as plt
```

## Pie chart of species percentages
Step 1: prepare the data

```python
count_by_species = pd.DataFrame(df.species.value_counts()).reset_index()
count_by_species
```

```python
fig = plt.figure(figsize=(8, 5))

plt.bar(x=count_by_species["species"], height=count_by_species["count"])

plt.savefig('my_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```

## Option 2: Matplotlib visuals built into pandas

```python
count_by_species.plot(kind="bar", x="species", y="count")
```

### Another example: Histplot

```python
fig = plt.Figure(figsize=(7,5))

plt.hist(df.sepal_length)
plt.savefig('basic_histogram.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig = plt.Figure(figsize=(7,5))

plt.hist(df.sepal_length, bins=20, color='skyblue', edgecolor='black')

plt.xlabel("Sepal Length (cm)")
plt.ylabel("Count")
plt.title("Distribution of iris flowers Sepal Lengths")
plt.grid(axis="y", alpha=0.75)

plt.savefig('stylized_histogram.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Assuming you have a DataFrame 'df' with columns 'sepal_length' and 'species'

fig, ax = plt.subplots(figsize=(7,5))

# Get unique species
species_list = df['species'].unique()

# Plot histogram for each species
for species in species_list:
    species_data = df[df['species'] == species]['sepal_length']
    ax.hist(species_data, bins=10, alpha=0.7, label=species, edgecolor='black')

ax.set_xlabel("Sepal Length (cm)")
ax.set_ylabel("Count")
ax.set_title("Distribution of Iris Flower Sepal Lengths by Species")
ax.grid(axis="y", alpha=0.75)
ax.legend()

plt.savefig('detailed_per_species_histogram.png', dpi=300, bbox_inches='tight')


plt.show()
```
