```python
import pandas as pd
df = pd.read_csv('IRIS.csv')
```

```python
df
```

```python
df.species.value_counts()
```

```python

setosa_df = df[df["species"] == "Iris-setosa"]

versicolor_df = df[df["species"] == "Iris-versicolor"]

virginica_df = df[df["species"] == "Iris-virginica"]
```

```python
virginica_df.species.value_counts()
```

```python
setosa_df.to_csv("setosa_data.csv", index=False)
versicolor_df.to_csv("versicolor_data.csv", index=False)
virginica_df.to_csv("virginica_data.csv", index=False)
```
