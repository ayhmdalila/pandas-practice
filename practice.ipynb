# Many operations you can do on dataframes:
## Indexing, calculations, plotting, joining etc...

### Indexing - selection

```python
# Selecting a specific column
df["species"]
```

```python
# Selecting rows (e.g. fifth row)
df.iloc[5]
```

```python
# Selecting multiple columns
df[["species", "Random Column"]]
```

```python
# Selecting a range of rows 
df.iloc[5:10]
```

```python
# Print the shape of a dataframe
df.shape
```

### Filtration

```python
# Filter for only species = "Iris-versicolor"
df[df["species"] == "Iris-versicolor"]
```

### Creating a new column out of an existing column

```python
df["Random * 10"] = df["Random Column"] * 10
```

```python
df
```
