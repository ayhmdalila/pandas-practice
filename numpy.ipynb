### Mathematics on arrays (broadcasting)

```python
x = np.random.rand(15)
y = np.random.rand(15)

print(x)
print("==")
print(y)
```

```python
z = []

for i in range(len(x)):
    z.append(x[i]+y[i])

return z
```

# Indexing in nD arrays

### Example: generate and plot a sine wave

```python
import numpy as np 
import matplotlib.pyplot as plt 
%matplotlib inline

# Generating the x axis
x = np.arange(0, 10, 0.1)
y = np.sin(x)
```

```python
fig = plt.Figure(figsize=(10,3))
plt.plot(y)

plt.show()
```

### Example 2: Scatter plot with random data

```python
x = np.random.randn(20)
y = np.random.randn(20)
sizes = 500 * np.random.rand(20)
```

```python
fig = plt.Figure(figsize=(10,10))
plt.scatter(x, y, s=sizes)
plt.title("Random X,Y points")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

### Example 3: Visualizing a normal distribution

```python
data = np.random.normal(loc=0, scale=1, size=100000)
```

```python
fig = plt.Figure(figsize=(10, 5))
plt.hist(data, bins=100)
plt.title("Histogram of Random Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
```
