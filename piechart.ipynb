```python
species_counts = df["species"].value_counts()
```

```python
plt.pie(
    species_counts,
    labels=species_counts.index,
    autopct="%1.1f%%"
)

plt.title("Iris Species Distribution")
plt.legend()

plt.savefig("pie_chart.png")
plt.show()
```
